# 03 — Time-frozen backtest (Tables 2 and 3)
Produced by `python -m backtest.run_backtest --freeze 2025-01-01`. Everything is as of the freeze date; the answer key is KEV additions after it.

```python
import json
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
import pandas as pd
bt = json.loads((ROOT / "backtest/results/backtest.json").read_text())
print("data:", bt["data_source"], "| freeze:", bt["freeze_date"], "| frozen model:", bt["frozen_model"]["algo"])
bt["headline"], bt["headline_population"]
```

```python
cols = ["name", "mean_rank_later_exploited", "precision_at_10", "precision_at_25", "effort_to_cover_90"]
for key, case in bt["cases"].items():
    print(f"\n== {key}: N={case['n_findings']}, later-exploited M={case['later_exploited']}, already in KEV at T={case['known_exploited_at_T']}")
    print(pd.DataFrame(case["table2"])[cols].set_index("name").round(3).to_string())
```

Open findings only (drops CVEs that were already in KEV at the freeze date):

```python
for key, case in bt["cases"].items():
    print(f"\n== {key}"); print(pd.DataFrame(case["table2_open_only"])[cols].set_index("name").round(3).to_string())
```
